```python
duplicate_count = df.duplicated().sum()

print("Duplicate Rows:", duplicate_count)
```

```python
df = df.drop_duplicates().copy()

print("Duplicates Remaining:", df.duplicated().sum())
```

```python
missing = pd.DataFrame({

    'Missing Count': df.isna().sum(),

    'Missing Percentage':
    (df.isna().mean() * 100).round(2)

})

missing = missing[
    missing['Missing Count'] > 0
]

missing.sort_values(
    'Missing Count',
    ascending=False
)
```

```python
categorical_columns = df.select_dtypes(
    include='object'
).columns

numerical_columns = df.select_dtypes(
    include=np.number
).columns

print("Categorical Columns:")

print(categorical_columns)

print("\nNumerical Columns:")

print(numerical_columns)
```

```python
for col in categorical_columns:

    df[col] = df[col].fillna('Unknown')
```

```python
for col in numerical_columns:

    df[col] = df[col].fillna(
        df[col].median()
    )
```

```python
season_summary = df.groupby('Season').agg(

    Average_Yield=('Yield_Tonnes_Ha', 'mean'),

    Average_Rainfall=('Rainfall_mm', 'mean'),

    Average_Temperature=('Avg_Temperature_C', 'mean'),

    Average_Water_Used=('Water_Used_m3', 'mean'),

    Average_Revenue=('Revenue_INR', 'mean'),

    Average_Profit=('Profit_INR', 'mean')

).round(2)

season_summary
```

```python
# Create all required seasonal averages

average_yield_season = df.groupby('Season')['Yield_Tonnes_Ha'].mean()

average_rainfall = df.groupby('Season')['Rainfall_mm'].mean()

average_water = df.groupby('Season')['Water_Used_m3'].mean()

average_profit = df.groupby('Season')['Profit_INR'].mean()


# Create Dashboard

fig, axes = plt.subplots(2, 2, figsize=(16, 10))

fig.suptitle(
    'Seasonal Agriculture Performance Dashboard',
    fontsize=20
)


# Graph 1: Average Yield by Season

average_yield_season.plot(
    kind='bar',
    ax=axes[0, 0]
)

axes[0, 0].set_title('Average Yield by Season')
axes[0, 0].set_xlabel('Season')
axes[0, 0].set_ylabel('Yield (Tonnes/Hectare)')


# Graph 2: Average Rainfall by Season

average_rainfall.plot(
    kind='bar',
    ax=axes[0, 1]
)

axes[0, 1].set_title('Average Rainfall by Season')
axes[0, 1].set_xlabel('Season')
axes[0, 1].set_ylabel('Rainfall (mm)')


# Graph 3: Average Water Usage by Season

average_water.plot(
    kind='bar',
    ax=axes[1, 0]
)

axes[1, 0].set_title('Average Water Usage by Season')
axes[1, 0].set_xlabel('Season')
axes[1, 0].set_ylabel('Water Used (m³)')


# Graph 4: Average Profit by Season

average_profit.plot(
    kind='bar',
    ax=axes[1, 1]
)

axes[1, 1].set_title('Average Profit by Season')
axes[1, 1].set_xlabel('Season')
axes[1, 1].set_ylabel('Profit (INR)')


# Adjust layout

plt.tight_layout(rect=[0, 0.03, 1, 0.95])

plt.show()
```
