```python
duplicate_count = df.duplicated().sum()

print("Duplicate Rows:", duplicate_count)
```

```python
df = df.drop_duplicates().copy()

print("Duplicates Remaining:", df.duplicated().sum())
```

```python
missing = pd.DataFrame({

    'Missing Count': df.isna().sum(),

    'Missing Percentage':
    (df.isna().mean() * 100).round(2)

})

missing = missing[
    missing['Missing Count'] > 0
]

missing.sort_values(
    'Missing Count',
    ascending=False
)
```

```python
categorical_columns = df.select_dtypes(
    include='object'
).columns

numerical_columns = df.select_dtypes(
    include=np.number
).columns

print("Categorical Columns:")

print(categorical_columns)

print("\nNumerical Columns:")

print(numerical_columns)
```

```python
for col in categorical_columns:

    df[col] = df[col].fillna('Unknown')
```

```python
for col in numerical_columns:

    df[col] = df[col].fillna(
        df[col].median()
    )
```

```python
print(df.isnull().sum())
```

```python
df['Season'].value_counts()
```

```python
season_counts = df['Season'].value_counts()

season_counts
```

```python
#Average Yield by Season
average_yield_season = df.groupby('Season')['Yield_Tonnes_Ha'].mean()

print(average_yield_season)

plt.figure(figsize=(8, 5))
average_yield_season.plot(kind='bar')

plt.title('Average Yield by Season')
plt.xlabel('Season')
plt.ylabel('Yield (Tonnes per Hectare)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
#Average Rainfall by Season
average_rainfall = df.groupby('Season')['Rainfall_mm'].mean()

print(average_rainfall)

plt.figure(figsize=(8, 5))
average_rainfall.plot(kind='bar')

plt.title('Average Rainfall by Season')
plt.xlabel('Season')
plt.ylabel('Rainfall (mm)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
#Average Water Usage by Season
average_water = df.groupby('Season')['Water_Used_m3'].mean()

print(average_water)

plt.figure(figsize=(8, 5))
average_water.plot(kind='bar')

plt.title('Average Water Usage by Season')
plt.xlabel('Season')
plt.ylabel('Water Used (m³)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
#Average Profit by Season
average_profit = df.groupby('Season')['Profit_INR'].mean()

print(average_profit)

plt.figure(figsize=(8, 5))
average_profit.plot(kind='bar')

plt.title('Average Profit by Season')
plt.xlabel('Season')
plt.ylabel('Profit (INR)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
#Crop Yield Across Seasons
plt.figure(figsize=(12, 6))

sns.barplot(
    data=df,
    x='Crop',
    y='Yield_Tonnes_Ha',
    hue='Season',
    errorbar=None
)

plt.title('Crop Yield Across Seasons')
plt.xlabel('Crop')
plt.ylabel('Yield (Tonnes per Hectare)')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
#Rainfall vs Crop Yield
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Rainfall_mm',
    y='Yield_Tonnes_Ha',
    hue='Season',
    alpha=0.6
)

plt.title('Rainfall vs Crop Yield')
plt.xlabel('Rainfall (mm)')
plt.ylabel('Yield (Tonnes per Hectare)')

plt.tight_layout()
plt.show()
```

```python
# Temperature vs Crop Yield
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Avg_Temperature_C',
    y='Yield_Tonnes_Ha',
    hue='Season',
    alpha=0.6
)

plt.title('Temperature vs Crop Yield')
plt.xlabel('Average Temperature (°C)')
plt.ylabel('Yield (Tonnes per Hectare)')

plt.tight_layout()
plt.show()
```

```python
#  Profit by Crop and Season
plt.figure(figsize=(12, 6))

sns.barplot(
    data=df,
    x='Crop',
    y='Profit_INR',
    hue='Season',
    errorbar=None
)

plt.title('Profit by Crop and Season')
plt.xlabel('Crop')
plt.ylabel('Profit (INR)')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
#Correlation Heatmap
numeric_df = df.select_dtypes(include=np.number)

correlation_matrix = numeric_df.corr()

plt.figure(figsize=(16, 10))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm'
)

plt.title('Correlation Heatmap of Agricultural Variables')

plt.tight_layout()
plt.show()
```

```python
#Seasonal Summary Table
season_summary = df.groupby('Season').agg(

    Average_Yield=('Yield_Tonnes_Ha', 'mean'),

    Average_Rainfall=('Rainfall_mm', 'mean'),

    Average_Temperature=('Avg_Temperature_C', 'mean'),

    Average_Water_Used=('Water_Used_m3', 'mean'),

    Average_Revenue=('Revenue_INR', 'mean'),

    Average_Profit=('Profit_INR', 'mean')

).round(2)

season_summary
```
